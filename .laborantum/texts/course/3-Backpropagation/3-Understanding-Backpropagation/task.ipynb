# Task 1

Write function that calculates value of this graph for any given input vector `x` and set of coefficients $b_1, b_2, c_1, c_2$ packed as vector of weights `w`.
In this function also return all the $z$ values.

````python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def dsigmoid(x):
    s = sigmoid(x)
    return s * (1 - s)


import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def graph_value(x, w):
    b1, b2, c1, c2 = w
    x1, x2 = x

    z1 = x1 + b1            
    z2 = x2 + b2            
    z3 = sigmoid(z1)        
    z4 = sigmoid(z2)        
    z5 = np.tanh(z2)        
    z6 = z5 * c2            
    z7 = z1 * z4            
    z8 = z7 * c1            
    z9 = z3 * z6            

    y  = z8 + z9            

    z = [z1, z2, z3, z4, z5, z6, z7, z8, z9]
    return y, z



x = [0.4967141530112327, -0.13826430117118466]
w = [0.6476885381006925, 1.5230298564080254, -0.23415337472333597, -0.23413695694918055]
y, z = graph_value(x, w)
print("y =", y)
print("z =", z)
````

````python
answer['graph_value'] = [graph_value(**input) for input in inputs]
````

# Task 2

In this graph, find all derivative of all the values over $z$ or $x$ or $w$.

Write the result in form of dictionary, for example, $\partial_{z_9} y$:

```
['dy_dz9'] = 1
```

````python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def graph_derivatives(x, w):
    b1, b2, c1, c2 = w
    x1, x2 = x

    z1 = x1 + b1
    z2 = x2 + b2
    z3 = sigmoid(z1)
    z4 = sigmoid(z2)
    z5 = np.tanh(z2)
    z6 = z5 * c2
    z7 = z1 * z4
    z8 = z7 * c1
    z9 = z3 * z6

    return {
        'dy_dz9': 1.0,
        'dy_dz8': 1.0,
        'dz9_dz6': float(z3),
        'dz9_dz3': float(z6),
        'dz8_dc1': float(z7),
        'dz8_dz7': float(c1),
        'dz7_dz1': float(z4),
        'dz7_dz4': float(z1),
        'dz6_dc2': float(z5),
        'dz6_dz5': float(c2),
        'dz5_dz2': float(1 - z5**2),
        'dz4_dz2': float(z4 * (1 - z4)),
        'dz3_dz1': float(z3 * (1 - z3)),
        'dz2_dx2': 1.0,
        'dz2_db2': 1.0,
        'dz1_dx1': 1.0,
        'dz1_db1': 1.0,
    }
````

````python
answer['graph_derivatives'] = [graph_derivatives(**input) for input in inputs]
````

# Task 3

Using the values of the derivatives, calculated above:
- extract a dictionary of values that are needed to calculate $\partial_{c_1} y$
- calculate that derivative

````python
def dy_dc1(x, w):
    b1, b2, c1, c2 = w
    x1, x2 = x
    z1 = x1 + b1
    z2 = x2 + b2
    z4 = 1 / (1 + np.exp(-z2))
    z7 = z1 * z4

    
    values = {
        'dz8_dc1': float(z7)
    }
    
    dy_dc11 = values['dz8_dc1']
    return values, dy_dc11
````

````python
answer['dy_dc1'] = [dy_dc1(**input) for input in inputs]
````

# Task 4
Using the values of the derivatives, calculated above:

````python
def dy_dc2(x, w):
    y, z = graph_value(x, w) 
    z3 = z[2]
    z5 = z[4]
    values = {
        'dz9_dz6': float(z3),
        'dz6_dc2': float(z5)
    }
    dy_dc22 = values['dz9_dz6'] * values['dz6_dc2']
    return values, dy_dc22

````

# Task 5
Using the values of the derivatives, calculated above:
- extract a dictionary of values that are needed to calculate $\partial_{b_1} y$
- calculate that derivative

````python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def dsigmoid(x):
    s = sigmoid(x)
    return s * (1 - s)

def dy_db1(x, w):
    b1, b2, c1, c2 = w
    x1, x2 = x
    z1 = x1 + b1
    z2 = x2 + b2
    z4 = 1 / (1 + np.exp(-z2))
    values = {
        'dy_dz8': 1.0,
        'dz8_dz7': float(c1),
        'dz7_dz1': float(z4),
        'dz1_db1': 1.0
    }
    dy_db11 = values['dy_dz8'] * values['dz8_dz7'] * values['dz7_dz1'] * values['dz1_db1']
    return values, dy_db11

````

````python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def dsigmoid(x):
    s = sigmoid(x)
    return s * (1 - s)

def dy_db2(x, w):
    b1, b2, c1, c2 = w
    x1, x2 = x
    z1 = x1 + b1
    z2 = x2 + b2
    z4 = 1 / (1 + np.exp(-z2))
    values = {
        'dy_dz8': 1.0,
        'dz8_dz7': float(c1),
        'dz7_dz4': float(z1),
        'dz4_dz2': float(z4 * (1 - z4)),
        'dz2_db2': 1.0
    }
    dy_db22 = values['dy_dz8'] * values['dz8_dz7'] * values['dz7_dz4'] * values['dz4_dz2'] * values['dz2_db2']
    return values, dy_db22
````
